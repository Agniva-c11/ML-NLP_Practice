```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(u"I will google about facebook")
doc.text
```

```python
doc[3]
```

```python
doc[3].pos_
```

```python
spacy.explain('ADP')
```

```python
doc[4].pos_
```

```python
spacy.explain('PROPN')
```

```python
for word in doc:
    print(word.text, "------>", word.pos_, spacy.explain(word.pos_))
```

```python
doc2 = nlp(u"I left the room")
for word in doc2:
    print(word.text, "------>", word.pos_, spacy.explain(word.pos_))
```

```python
doc3 = nlp(u"to the left of the room")
for word in doc3:
    print(word.text, "------>", word.pos_, spacy.explain(word.pos_))
```
