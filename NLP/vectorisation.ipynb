#### Bag of Words

```python
import numpy as np
import pandas as pd
df = pd.DataFrame({'text':['people watch campusx', 'campusx watch campusx', 'people write comment', 'campusx write comment'], 'output':[1, 1, 0, 0]})
```

```python
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(ngram_range=(1, 2))
```

```python
bow = cv.fit_transform(df['text'])
```

```python
# vocabulary
print(cv.vocabulary_)
```

```python
print(bow[0].toarray())
print(bow[1].toarray())
print(bow[2].toarray())
print(bow[3].toarray())
```

```python
# New sentence
cv.transform(["campusx watch and write campusx of campusx"]).toarray()
```

#### Tfidf vectorisation


```python
import numpy as np
import pandas as pd
df1 = pd.DataFrame({'text':['people watch campusx', 'campusx watch campusx', 'people write comment', 'campusx write comment'], 'output':[1, 1, 0, 0]})
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()
tfidf.fit_transform(df1['text']).toarray()
```

```python
# idf values
print(tfidf.idf_)

# vocabulary
print(tfidf.get_feature_names_out())
```
